```python
import pandas as pd

df_raw = pd.read_csv('temp_dataset.csv')

raw_path = os.path.join(BASE, 'data/raw/Telco-Customer-Churn.csv')
df_raw.to_csv(raw_path, index=False)

df = df_raw.copy()
print("Shape:", df.shape)
df.head()
```

```python
print(df.shape)
df.info()
```

```python
for col in df.columns:
    print(f"{col}: {df[col].nunique()} unique values")
```

```python
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
missing = df.isna().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_summary = pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
missing_summary[missing_summary['missing_count'] > 0]
```

```python
print("Full duplicate rows:", df.duplicated().sum())
print("Duplicate customerID:", df['customerID'].duplicated().sum())
```

```python
df.describe()
```

```python
blank_mask = df['TotalCharges'].str.strip() == ''
print("Blank TotalCharges rows:", blank_mask.sum())
df.loc[blank_mask, ['customerID', 'tenure', 'TotalCharges']]
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

df['TotalCharges'] = df['TotalCharges'].fillna(0)

print(df['TotalCharges'].dtype)
print("Remaining NaNs:", df['TotalCharges'].isna().sum())
```

```python
df['SeniorCitizen'] = df['SeniorCitizen'].map({0: 'No', 1: 'Yes'})
df['SeniorCitizen'].value_counts()
```

```python
customer_ids = df['customerID'].copy()

df_model = df.drop(columns=['customerID'])
print(df_model.shape)
```

```python
for col in df_model.select_dtypes(include='object').columns:
    df_model[col] = df_model[col].str.strip()
```

```python
data_dict = pd.DataFrame({
    'column': df_model.columns,
    'dtype': df_model.dtypes.astype(str).values,
    'n_unique': [df_model[c].nunique() for c in df_model.columns],
    'sample_values': [df_model[c].dropna().unique()[:3].tolist() for c in df_model.columns]
})

dict_path = os.path.join(BASE, 'data/processed/data_dictionary.csv')
data_dict.to_csv(dict_path, index=False)
data_dict
```

```python
cleaning_log = """
CLEANING LOG — Telco Customer Churn
-----------------------------------
1. Loaded raw file from Google Drive link (file ID: 1p50C3kU-Vla809qMeQNcEiMRvqJk8AJH)
   and stored untouched copy at data/raw/Telco-Customer-Churn.csv (shape: {raw_shape}).
2. Checked for full-row duplicates: {dup_count} found.
3. Checked for duplicate customerID: {dup_id_count} found.
4. TotalCharges: originally object dtype due to {blank_count} blank-string entries,
   all corresponding to tenure = 0. Converted via pd.to_numeric(errors='coerce'),
   then filled resulting NaNs with 0 (justification: zero tenure means no charges
   have accrued yet — factually correct, not an imputed guess).
5. SeniorCitizen: originally int64 (0/1), remapped to categorical 'No'/'Yes' for
   readability and consistency with other categorical columns.
6. customerID: dropped from modeling dataframe (df_model) — unique identifier,
   no predictive value. Retained separately in `customer_ids` for reporting/joins.
7. Stripped leading/trailing whitespace from all categorical (object) columns.
8. No missing values remain after treatment.
9. Final processed shape: {final_shape}
""".format(
    raw_shape=df_raw.shape,
    dup_count=df.duplicated().sum(),
    dup_id_count=df['customerID'].duplicated().sum(),
    blank_count=blank_mask.sum(),
    final_shape=df_model.shape
)

log_path = os.path.join(BASE, 'data/processed/cleaning_log.txt')
with open(log_path, 'w') as f:
    f.write(cleaning_log)

print(cleaning_log)
```
